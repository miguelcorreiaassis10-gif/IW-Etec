```python
n1 = int(input("digite um numero "))
n2 = int(input("digite um numero "))
if n1 == n2:
  print(" os dois são iguais ")
else:
  print(" os dois são diferentes")
```

```python
nota = float(input("digite sua nota media "))
if nota >= 7:
  print(" aprovado")
elif nota >= 5 and nota <= 7:
  print("Recuperação")
else:
  print("reprovado")
```

```python
valor = float(input(" qual o valor da compra "))
if valor >= 100:
  desconto = valor / 10
  vf = valor - desconto
  print(f"valor final com desconto: {vf}")
else:
  print(f"valor da compra é {valor}")
```

```python
n = float(input("digite um numero "))
if n >= 0:
  print(" o numero é positivo")
else:
  print("o numero é negativo")
```

```python
idade = int(input("qual sua idade? "))
if idade <= 12:
  print("criança")
elif idade <= 18 and idade >= 13:
  print("adolescente")
elif idade >= 19 and idade <= 59:
  print("adulto")
else:
  print("idoso")
```

```python
nota1 = float(input("digite sua nota  "))
nota2 = float(input("digite sua nota  "))
notab = (nota1 + nota2) / 2
if notab >= 7:
  print(" aprovado")
else:
  print("reprovado")
```

```python
usuario = str(input("digite seu nome de usuario "))
senha = int(input("digite sua senha"))
print("cadastro realizado")
lusuario = str(input("digite seu nome de usuario "))
lsenha = int(input("digite sua senha"))
if usuario == lusuario and senha == lsenha:
  print("acesso permitido")
else:
  print(" acesso negado")
```

```python
peso = float(input("qual seu peso? "))
altura = float(input("qual sua altura?"))
imc = peso / (altura * altura)
if imc < 18.5:
  print("abaixo do peso")
elif imc >= 18.5 and imc <= 24.9:
  print("peso normal")
elif imc >= 25 and imc <= 29.9:
  print("sobrepeso")
elif imc >= 30:
  print("obeso")
```
