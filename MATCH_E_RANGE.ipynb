MATCH

```python
cor = "azul"

match cor:
    case "azul":
        print("Cor azul")
    case _:
        print("Outra cor")
```

```python
numero = 2

match numero:
    case 1:
        print("Um")
    case 2:
        print("Dois")
    case 3:
        print("Três")
```

```python
opcao = 1

match opcao:
    case 1:
        print("Água")
    case 2:
        print("Suco")
    case 3:
        print("Refrigerante")
    case _:
        print("Opção inválida")
```

```python
dia = "sábado"

match dia:
    case "sábado" | "domingo":
        print("Final de semana")
    case _:
        print("Dia de semana")
```

```python
nota = "A"

match nota:
    case "A":
        print("Excelente")
    case "B":
        print("Bom")
    case "C":
        print("Precisa melhorar")
    case _:
        print("Nota inválida")
```

RANGE

```python
for i in range(10):
    print(i)
```

```python
for i in range(1, 11):
    print(i)
```

```python
for i in range(5, 16):
    print(i)
```

```python
for i in range(0, 21, 2):
    print(i)
```

```python
for i in range(1, 16, 2):
    print(i)
```

```python
for i in range(10, 0, -1):
    print(i)
```

```python
for i in range(1, 11):
    print("3 x", i, "=", 3 * i)
```
