# *dicionario*

```python
aluno = {
    "nome": "Lucas",
    "idade": 20,
    "curso": "Engenharia"
}
print(aluno)
```

```python
aluno = {
    "nome": "Mariana",
    "idade": 22,
    "curso": "Direito"
}

print(aluno["nome"])
print(aluno["idade"])
print(aluno["curso"])
```

```python
produto = {
    "nome": "Telemóvel",
    "preço": 1500.0,
    "estoque": 10
}

# Alterando o preço
produto["preço"] = 1350.0

print(produto)
```

```python
aluno = {
    "nome": "Pedro",
    "idade": 19
}

# Adicionando a informação curso
aluno["curso"] = "Informática"

print(aluno)
```

```python
filme = {
    "título": "Inception",
    "ano": 2010,
    "género": "Ficção Científica",
    "nota": 8.8
}

print(filme["título"])
print(filme["ano"])
print(filme["género"])
print(filme["nota"])
```

# **CONJUNTO**

```python
numeros = {1, 2, 3, 4, 5}
print(numeros)
```

```python
numeros = {1, 2, 2, 3, 3, 3, 4, 5}
print(numeros)
```

```python
frutas = {"maçã", "banana", "laranja"}
frutas.add("morango")
print(frutas)
```

```python
cores = {"vermelho", "azul", "verde", "amarelo"}
cores.remove("verde")
print(cores)
```

```python
nomes = ["Yris", "Luna", "Luna", "Maitê", "Yris", "Ravi", "Gael", "Theo", "Theo"]

# Transformando a lista num conjunto
nomes_unicos = set(nomes)

print(nomes_unicos)
```
