```python
matriz = [
    [3, 6, 9],
    [12, 15, 18],
    [21, 24, 27]
]

coluna = []

for linha in matriz:
    coluna.append(linha[1])

print(coluna)
```

```python
matriz = [
    [1, 2],
    [3, 4],
    [5, 6]
]

for linha in matriz:
    for numero in linha:
        print(numero)
```

```python
matriz = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

soma = 0

for linha in matriz:
    soma += linha[1]

print(soma)
```

```python
matriz = [
    [12, 5, 8],
    [20, 3, 15],
    [7, 25, 10]
]

maior = matriz[0][0]

for linha in matriz:
    for numero in linha:
        if numero > maior:
            maior = numero

print(maior)
```

```python
notas = [
    [7, 8, 9],
    [6, 5, 8],
    [10, 9, 7]
]

# a) Notas do primeiro aluno
print(notas[0])

# b) Segunda nota do segundo aluno
print(notas[1][1])

# c) Terceira nota de cada aluno
for aluno in notas:
    print(aluno[2])

# d) Alterar a primeira nota do terceiro aluno
notas[2][0] = 8
print(notas)
```
